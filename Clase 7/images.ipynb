```python
import cv2
import urllib
import numpy as np
import matplotlib.pyplot as plt

req = urllib.request.urlopen('https://content.elmueble.com/medio/2023/05/22/gatito_c529636e_230522123719_1000x667.jpg')
arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
img = cv2.imdecode(arr, -1)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.xlabel("filas")
plt.imshow(img_rgb)
plt.show()
```

```python
img_rgb.shape
```

```python
b, g, r = cv2.split(img_rgb)
```

```python
img1_b_gray = b.copy()
img1_g_gray = np.zeros(g.shape, dtype = np.uint8)
img1_r_gray = np.zeros(r.shape, dtype = np.uint8)
```

```python
img1_g_gray
```

```python
img1_gray = cv2.merge((img1_b_gray, img1_g_gray, img1_r_gray))
```

```python

img1_gray = cv2.cvtColor(img1_gray, cv2.COLOR_BGR2GRAY)

plt.xlabel("filas")
plt.imshow(img1_gray, cmap="gray")
plt.show()
```

```python
img2_b_gray = np.zeros(b.shape, dtype = np.uint8) 
img2_g_gray = g.copy()
img2_r_gray = np.zeros(r.shape, dtype = np.uint8)
img2_gray = cv2.merge((img2_b_gray, img2_g_gray, img2_r_gray))

img2_gray = cv2.cvtColor(img2_gray, cv2.COLOR_BGR2GRAY)

plt.xlabel("filas")
plt.imshow(img2_gray, cmap="gray")
plt.show()
```

```python
img3_b_gray = np.zeros(b.shape, dtype = np.uint8) 
img3_g_gray =  np.zeros(g.shape, dtype = np.uint8)
img3_r_gray = r.copy()
img3_gray = cv2.merge((img3_b_gray, img3_g_gray, img3_r_gray))

img3_gray = cv2.cvtColor(img3_gray, cv2.COLOR_BGR2GRAY)

plt.xlabel("filas")
plt.imshow(img3_gray, cmap="gray")
plt.show()
```

```python
canal_1 = (b + g + r) // 3
canal_2 = np.zeros(b.shape, dtype = np.uint8) 
canal_3 = np.zeros(b.shape, dtype = np.uint8) 
```

```python
img_mean_gray = cv2.merge((canal_1, canal_2, canal_3))
img_mean_gray = cv2.cvtColor(img_mean_gray, cv2.COLOR_BGR2GRAY)

plt.xlabel("filas")
plt.imshow(img_mean_gray, cmap="gray")
plt.show()
```
