```python
import random
```

```python
ITERACIONES = 10000000
```

```python
random.random() * 2
```

```python
resultados = []
for i in range(ITERACIONES):
    x = random.random() * 2 - 1
    y = random.random() * 2 - 1
    resultados.append((x ** 2 + y ** 2) <= 1)
```

```python
pi = 4 * sum(resultados) / ITERACIONES
pi
```
